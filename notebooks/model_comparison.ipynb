```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "XGBoost"
    ],
    "Accuracy": [
        0.9470,
        0.9855,
        0.9830
    ],
    "Precision": [
        0.9116,
        0.9812,
        0.9773
    ],
    "Recall": [
        0.9900,
        0.9900,
        0.9890
    ],
    "F1-Score": [
        0.9492,
        0.9856,
        0.9831
    ],
    "ROC-AUC": [
        0.9858,
        0.9952,
        0.9931
    ]
})
```

```python
display(results)
```

```python
plot_data = results.set_index("Model").T

# Create grouped bar chart
plt.figure(figsize=(12, 6))

plot_data.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Machine Learning Model Performance Comparison", fontsize=16)
plt.xlabel("Evaluation Metric", fontsize=12)
plt.ylabel("Score", fontsize=12)

plt.ylim(0.85, 1.01)

plt.xticks(rotation=0)
plt.legend(title="Model")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
